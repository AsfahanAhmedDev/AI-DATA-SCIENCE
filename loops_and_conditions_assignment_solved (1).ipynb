## ⚖️ 2. BMI Health Checker

### Goal:
    Help a fitness app tell users which weight category they belong to.

### Steps:

    Ask for weight in kg and height in metres.
    
    Calculate BMI = weight / (height * height).
    
    BMI below 18.5 → "Underweight".
    
    18.5 to 24.9 → "Normal".
    
    25 to 29.9 → "Overweight".
    
    30 or more → "Obese".
    
    Print the BMI rounded to 2 decimals.

### Example:

    Enter weight: 70
    Enter height: 1.75
    BMI = 22.86 → Normal

```python
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in metres: "))
if height <= 0 or weight <= 0:
    print("Invalid weight or height")
else:
    bmi = weight / (height * height)
    print("BMI:", round(bmi, 2))
    if bmi < 18.5:
        print("Underweight")
    elif bmi <= 24.9:
        print("Normal")
    elif bmi <= 29.9:
        print("Overweight")
    else:
        print("Obese")
```

## ✊ 17. Rock, Paper, Scissors Game

### Goal:
    Decide the winner of a two-player game.

### Steps:

    Ask Player 1 and Player 2 for rock, paper or scissors.
    
    Rock beats scissors, scissors beats paper, paper beats rock.
    
    Use logical operators (and / or) to print the winner.
    
    If both choose the same → "Draw".

### Example:

    Player 1: rock
    Player 2: scissors
    Player 1 wins

```python
p1 = input("Player 1 (rock/paper/scissors): ").strip().lower()
p2 = input("Player 2 (rock/paper/scissors): ").strip().lower()
valid = ["rock", "paper", "scissors"]
if p1 not in valid or p2 not in valid:
    print("Invalid choice")
elif p1 == p2:
    print("Draw")
elif (p1 == "rock" and p2 == "scissors") or (p1 == "scissors" and p2 == "paper") or (p1 == "paper" and p2 == "rock"):
    print("Player 1 wins")
else:
    print("Player 2 wins")
```

## 🌡️ 50. Weekly Temperature Report

### Goal:
    A weather station wants to know how many days were warmer than average.

### Steps:

    Create temps = [30, 32, 28, 35, 31, 29, 36].
    
    First loop: add all the temperatures and calculate the average.
    
    Second loop: count the days above the average.
    
    Print the average (2 decimals) and the count.

### Example:

    Average = 31.57
    Days above average = 3

```python
temps = [30, 32, 28, 35, 31, 29, 36]
total = 0
for temp in temps:
    total += temp
average = total / len(temps)
count = 0
for temp in temps:
    if temp > average:
        count += 1
print("Average temperature:", round(average, 2))
print("Days above average:", count)
```
